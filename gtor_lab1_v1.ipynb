## Задание 1. Решить задачу линейного программирования перебором вершин

```python
A = np.array([
    [3, 1],
    [2, 6],
    [-1, 0],
    [0, -1]
], dtype="float")

b = np.array([15, 8, 0, 0], dtype="float")

c = np.array([7, 2], dtype="float")

allowed, costs = helpers.solve_traversal(A, b, c)

for idx, (sol, cost) in enumerate(zip(allowed, costs)):
    print(f"Решение: {abs(sol).round(2)} \t| Целевая функция: {cost:.2f}")

best_solution = abs(allowed[0])
best_cost = costs[0]
print(f"\nЛучшее решение: {best_solution}")
print(f"Максимальное значение целевой функции: {best_cost}")
```
